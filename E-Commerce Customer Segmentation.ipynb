## E-Commerce Customer Segmentation

DATASET DETAILS:

- InvoiceNo: Invoice number. Nominal. A 6-digit integral number uniquely assigned to each transaction. If this code starts with the letter 'c', it indicates a cancellation.
- StockCode: Product (item) code. Nominal. A 5-digit integral number uniquely assigned to each distinct product.
- Description: Product (item) name. Nominal.
- Quantity: The quantities of each product (item) per transaction. Numeric.
- InvoiceDate: Invice date and time. Numeric. The day and time when a transaction was generated.
- UnitPrice: Unit price. Numeric. Product price per unit in sterling (Â£).
- CustomerID: Customer number. Nominal. A 5-digit integral number uniquely assigned to each customer.
- Country: Country name. Nominal. The name of the country where a customer resides.

## RFM 

```python
df_clean.columns
```

### Recency

```python
customer_last_purchase = df_clean.groupby('Customer ID')['InvoiceDate'].max()
customer_last_purchase.head()
```

```python
last_date = df_clean['InvoiceDate'].max()
last_date
```

```python
recency = (last_date - customer_last_purchase).dt.days
recency
```

### Frequency

```python
frequency= df_clean.groupby('Customer ID')['Invoice'].nunique()
frequency.head()
```

### Monetary

```python
df_clean['TotalAmount']= df_clean['Quantity']*df_clean['Price']
```

```python
monetary = df_clean.groupby('Customer ID')['TotalAmount'].sum()
monetary
```

```python
rfm = pd.concat([recency,frequency,monetary],axis=1)
```

```python
rfm.columns = ['Recency','Frequency','Monetary']
rfm
```

```python
rfm['R_Score'] = pd.qcut(rfm['Recency'],5,labels=[5, 4, 3, 2, 1])
```

```python
rfm['F_Score'] = pd.qcut(rfm['Frequency'].rank(method='first'),5,labels=[1,2,3,4,5])
```

```python
rfm['M_Score'] = pd.qcut(rfm['Monetary'],5,labels=[1,2,3,4,5])
```

```python
rfm['RFM_Total'] = ( rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str) )
```

```python
rfm['RFM_Score'] = ( rfm['R_Score'].astype(int) + rfm['F_Score'].astype(int) + rfm['M_Score'].astype(int) )
```

```python
rfm["Customer_segment"] = np.where(rfm['RFM_Score'] >= 12, "Top Customers",
                          np.where(rfm['RFM_Score'] >= 9, "High Value Customer",
                          np.where(rfm['RFM_Score'] >= 6, "Medium Value Customer",
                          "Low Value Customers")))
```
